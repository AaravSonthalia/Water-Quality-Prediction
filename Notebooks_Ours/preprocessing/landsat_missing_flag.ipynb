# Landsat Missing Flag

Adds a `landsat_present` flag to all six Final Datasets, keeping NaNs (no imputation).

```python
import os
import pandas as pd
from pathlib import Path


def resolve_project_root() -> str:
    cwd = os.getcwd()
    cwd_base = os.path.basename(cwd)
    if cwd_base == "preprocessing":
        return os.path.abspath(os.path.join(cwd, "..", ".."))
    if cwd_base == "Notebooks_Ours":
        return os.path.abspath(os.path.join(cwd, ".."))
    return os.path.abspath(cwd)


PROJECT_ROOT = resolve_project_root()
FINAL_DIR = Path(PROJECT_ROOT) / "Datasets_Ours" / "Final Datasets"

FILES = [
    "drp_training_complete.csv",
    "ec_training_complete.csv",
    "ta_training_complete.csv",
    "drp_validation.csv",
    "ec_validation.csv",
    "ta_validation.csv",
]

# Core Landsat bands used to determine presence
LANDSAT_COLS = ["nir", "ndmi", "blue", "red", "swir22", "swir16", "green", "mndwi"]

print("Adding landsat_present flag (no imputation)...")

for fname in FILES:
    path = FINAL_DIR / fname
    df = pd.read_csv(path)

    cols = [c for c in LANDSAT_COLS if c in df.columns]
    if not cols:
        print(f"{fname}: no Landsat columns found, skipping")
        continue

    df["landsat_present"] = (~df[cols].isna().any(axis=1)).astype(int)
    df.to_csv(path, index=False)
    print(f"{fname}: landsat_present added (using {len(cols)} bands)")

print("✓ Done.")
```
